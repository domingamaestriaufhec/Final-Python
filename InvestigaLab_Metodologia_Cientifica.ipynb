## 📐 1. Nivel 3: Cálculo del Tamaño de Muestra para Poblaciones Finitas
Fórmula estadística clásica para estimar proporciones con población finita:

$$n = \frac{N \cdot Z^2 \cdot p \cdot q}{e^2 \cdot (N - 1) + Z^2 \cdot p \cdot q}$$


```python
def calcular_muestra_finita(N, Z=1.96, p=0.5, e=0.05):
    """Calcula el tamaño muestral representativo para una población finita."""
    q = 1 - p
    numerador = N * (Z**2) * p * q
    denominador = (e**2) * (N - 1) + (Z**2) * p * q
    n = numerador / denominador
    return round(n)

# Ejemplo: Facultad universitaria de 2,400 estudiantes
N_poblacion = 2400
muestra = calcular_muestra_finita(N_poblacion, Z=1.96, p=0.5, e=0.05)
print(f"Para una población N = {N_poblacion} (95% confianza, 5% error):")
print(f"-> Tamaño de Muestra Requerido (n) = {muestra} sujetos")
```

## 📊 2. Nivel 4: Confiabilidad por Consistencia Interna (Alfa de Cronbach)
Fórmula psicométrica de Cronbach:

$$\alpha = \frac{K}{K - 1} \left( 1 - \frac{\sum s_i^2}{s_t^2} \right)$$


```python
import numpy as np

def alfa_cronbach(matriz_respuestas):
    """Calcula el coeficiente Alfa de Cronbach para una matriz de encuestas (sujetos x ítems)."""
    matriz = np.array(matriz_respuestas)
    K = matriz.shape[1] # Número de ítems
    varianzas_items = np.var(matriz, axis=0, ddof=1)
    suma_varianzas = np.sum(varianzas_items)
    puntajes_totales = np.sum(matriz, axis=1)
    varianza_total = np.var(puntajes_totales, ddof=1)
    alfa = (K / (K - 1)) * (1 - (suma_varianzas / varianza_total))
    return alfa

# Simulación de respuestas de 10 estudiantes a un cuestionario de 5 ítems Likert (1 a 5)
np.random.seed(42)
datos_encuesta = np.random.randint(3, 6, size=(10, 5))
alfa = alfa_cronbach(datos_encuesta)
print(f"Matriz de Respuestas (10 sujetos x 5 ítems):\n{datos_encuesta}\n")
print(f"-> Coeficiente Alfa de Cronbach: {alfa:.4f}")
print("Interpretación:", "Excelente / Confiable" if alfa >= 0.80 else "Aceptable")
```

## 🧪 3. Nivel 4: Tamaño del Efecto (d de Cohen)
Cuantifica la magnitud o impacto práctico de una intervención pedagógica:

$$d = \frac{\bar{X}_1 - \bar{X}_2}{s_{\text{agrupada}}}$$


```python
def d_cohen(grupo1, grupo2):
    """Calcula el tamaño del efecto d de Cohen entre dos grupos independientes."""
    m1, m2 = np.mean(grupo1), np.mean(grupo2)
    s1, s2 = np.var(grupo1, ddof=1), np.var(grupo2, ddof=1)
    n1, n2 = len(grupo1), len(grupo2)
    s_agrupada = np.sqrt(((n1 - 1) * s1 + (n2 - 1) * s2) / (n1 + n2 - 2))
    return (m1 - m2) / s_agrupada

# Calificaciones postest (Grupo Experimental vs Grupo Control)
grupo_exp = [17, 18, 19, 16, 18, 20, 19, 17, 18, 19]
grupo_ctrl = [13, 14, 12, 15, 13, 14, 13, 12, 14, 13]

d = d_cohen(grupo_exp, grupo_ctrl)
print(f"Media Grupo Experimental: {np.mean(grupo_exp):.2f}")
print(f"Media Grupo Control: {np.mean(grupo_ctrl):.2f}")
print(f"-> Tamaño del Efecto (d de Cohen) = {d:.2f}")
print("Magnitud:", "Efecto Grande (d >= 0.80)" if d >= 0.80 else "Efecto Moderado")
```
